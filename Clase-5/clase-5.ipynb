# Solarizacion o Posterizacion

SOLARIZACIÓN

```python
import cv2 
import numpy as np
import matplotlib.pyplot as plt

def solarizar(imagen, umbral):
    # Crear una copia de la imagen para modificarla
    imagen_sol = np.where(imagen < umbral, imagen, 255 - imagen)
    return imagen_sol
    
imagen = cv2.imread('colores2.jpg')  # Cargar la imagen en color 
img_gray=cv2.cvtColor(imagen, cv2.COLOR_BGR2GRAY)  # Convertir a escala de grises


umbral = 127  # Definir el umbral para la solarización
img_fin=solarizar(img_gray, umbral)  # Aplicar la función de solarización


plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.title('Imagen Original')
plt.imshow(cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB))  # Mostrar la imagen original

plt.subplot(1, 2, 2)
plt.title('Imagen Solarizada en escala de grises')
plt.imshow(img_fin, cmap='gray')  # Mostrar la imagen solarizada en escala de grises
plt.show()
```

POSTERIZACION

```python
import cv2 
import numpy as np
import matplotlib.pyplot as plt

def posterizar(imagen,niveles):
    factor_escala=255/niveles
    img_pos=np.round(imagen/factor_escala)*factor_escala
    return img_pos.astype(np.uint8)

imagen = cv2.imread('colores2.jpg')  # Cargar la imagen en color 
img_gray=cv2.cvtColor(imagen, cv2.COLOR_BGR2GRAY)  # Convertir a escala de grises   
nvl=3
img_fin=posterizar(img_gray, nvl)  # Aplicar la función de solarización


plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.title('Imagen Original')
plt.imshow(cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB)) 
plt.subplot(1, 2, 2)
plt.title('Imagen Posterizada')
plt.imshow(img_fin, cmap='gray') 
plt.show() 
```

```python
import cv2 
import numpy as np
import matplotlib.pyplot as plt

def posterizar(imagen,niveles):
    factor_escala=255/niveles
    img_pos=np.round(imagen/factor_escala)*factor_escala
    return img_pos.astype(np.uint8)

def solarizar(imagen, umbral):
    # Crear una copia de la imagen para modificarla
    imagen_sol = np.where(imagen < umbral, imagen, 255 - imagen)
    return imagen_sol

imagen = cv2.imread('colores2.jpg')  
nvl=3
img_fin=posterizar(imagen, nvl)  
umbral = 127  # Definir el umbral para la solarización
img_fin2=solarizar(imagen, umbral) 

plt.figure(figsize=(10, 5))
plt.subplot(1, 3, 1)
plt.title('Imagen Original')
plt.imshow(cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB))  # Mostrar la imagen original

plt.subplot(1, 3, 2)
plt.title('Imagen Solarizada')
plt.imshow(cv2.cvtColor(img_fin2, cv2.COLOR_BGR2RGB)) 

plt.subplot(1, 3, 3)
plt.title('Imagen Posterisada')
plt.imshow(cv2.cvtColor(img_fin, cv2.COLOR_BGR2RGB)) 
plt.show() 
```

FALSO COLOR

```python
import cv2 
import numpy as np
import matplotlib.pyplot as plt

def falso_color(img_gray):
    escal_img=cv2.normalize(img_gray,None,0,255,cv2.NORM_MINMAX)
    falso_img=cv2.applyColorMap(escal_img,cv2.COLORMAP_JET)
    return falso_img

imagen = cv2.imread('colores2.jpg')  # Cargar la imagen en color 
img_gray=cv2.cvtColor(imagen, cv2.COLOR_BGR2GRAY) 
img_fin=falso_color(img_gray)   

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 2)
plt.title('Imagen Falso Color')
plt.imshow(cv2.cvtColor(img_fin, cv2.COLOR_BGR2RGB))

plt.subplot(1, 2, 1)
plt.title('Imagen Original')
plt.imshow(img_gray, cmap='gray') 
plt.show() 
```

Falso Color Con Todas las Paletas

```python
import cv2 
import numpy as np
import matplotlib.pyplot as plt

def falso_color(img_gray, colormap):
    escal_img = cv2.normalize(img_gray, None, 0, 255, cv2.NORM_MINMAX)
    falso_img = cv2.applyColorMap(escal_img, colormap)
    return falso_img

# Leer imagen
imagen = cv2.imread('colores2.jpg')

if imagen is None:
    print("Error al cargar la imagen")
    exit()

# Convertir a gris
img_gray = cv2.cvtColor(imagen, cv2.COLOR_BGR2GRAY)
# Arreglo de paletas
colormaps = [
    ("AUTUMN", cv2.COLORMAP_AUTUMN),
    ("BONE", cv2.COLORMAP_BONE),
    ("JET", cv2.COLORMAP_JET),
    ("WINTER", cv2.COLORMAP_WINTER),
    ("RAINBOW", cv2.COLORMAP_RAINBOW),
    ("OCEAN", cv2.COLORMAP_OCEAN),
    ("SUMMER", cv2.COLORMAP_SUMMER),
    ("SPRING", cv2.COLORMAP_SPRING),
    ("COOL", cv2.COLORMAP_COOL),
    ("HSV", cv2.COLORMAP_HSV),
    ("PINK", cv2.COLORMAP_PINK),
    ("HOT", cv2.COLORMAP_HOT),
    ("PARULA", cv2.COLORMAP_PARULA),
    ("MAGMA", cv2.COLORMAP_MAGMA),
    ("INFERNO", cv2.COLORMAP_INFERNO),
    ("PLASMA", cv2.COLORMAP_PLASMA),
    ("VIRIDIS", cv2.COLORMAP_VIRIDIS),
    ("CIVIDIS", cv2.COLORMAP_CIVIDIS),
    ("TWILIGHT", cv2.COLORMAP_TWILIGHT),
    ("TWILIGHT_SHIFTED", cv2.COLORMAP_TWILIGHT_SHIFTED),
    ("TURBO", cv2.COLORMAP_TURBO),
    ("DEEPGREEN", cv2.COLORMAP_DEEPGREEN)
]
# Creacion de filas y columnas
num_maps = len(colormaps)
cols = 4
rows = int(np.ceil(num_maps / cols))
# Creacion de la figura
plt.figure(figsize=(15, rows * 3))

# Imprecion de las paletas
for i, (nombre, cmap) in enumerate(colormaps):
    img_color = falso_color(img_gray, cmap)

    plt.subplot(rows, cols, i + 1)
    plt.title(nombre)
    plt.imshow(cv2.cvtColor(img_color, cv2.COLOR_BGR2RGB))
    plt.axis('off')

plt.tight_layout()
plt.show()
```
