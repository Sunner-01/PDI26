RGB usando matplotlib

```python
import cv2
import matplotlib.pyplot as plt

plt.rcParams['figure.figsize'] = [10,5]

# Cargar imágenes
img1 = cv2.imread('foto-1.webp')
img2 = cv2.imread('foto-2.webp')
img3 = cv2.imread('foto-3.webp')

# Convertir de BGR a RGB
img1_rgb = cv2.cvtColor(img1, cv2.COLOR_BGR2RGB)
img2_rgb = cv2.cvtColor(img2, cv2.COLOR_BGR2RGB)
img3_rgb = cv2.cvtColor(img3, cv2.COLOR_BGR2RGB)

# Mostrar imágenes
plt.figure()
plt.imshow(img1_rgb)
plt.title("Imagen 1")
plt.axis("off")
plt.show()

plt.figure()
plt.imshow(img2_rgb)
plt.title("Imagen 2")
plt.axis("off")
plt.show()

plt.figure()
plt.imshow(img3_rgb)
plt.title("Imagen 3")
plt.axis("off")
plt.show()
```

Escala de grises con cv2

```python
import cv2
import matplotlib.pyplot as plt

plt.rcParams['figure.figsize'] = [15,5]

# Cargar imagenes en BGR
img1 = cv2.imread('foto-1.webp')
img2 = cv2.imread('foto-2.webp')
img3 = cv2.imread('foto-3.webp')

# Convertir a escala de grises
gray1 = cv2.cvtColor(img1, cv2.COLOR_BGR2GRAY)
gray2 = cv2.cvtColor(img2, cv2.COLOR_BGR2GRAY)
gray3 = cv2.cvtColor(img3, cv2.COLOR_BGR2GRAY)

plt.figure()

plt.subplot(1,3,1)
plt.imshow(gray1, cmap='gray')
plt.title("Imagen 1")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(gray2, cmap='gray')
plt.title("Imagen 2")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(gray3, cmap='gray')
plt.title("Imagen 3")
plt.axis("off")

plt.tight_layout()
plt.show()
```

Guardar Grises

```python
cv2.imwrite('foto-1_gris.jpg', gray1)
cv2.imwrite('foto-2_gris.jpg', gray2)
cv2.imwrite('foto-3_gris.jpg', gray3)
```

Función

```python
def mostrar_imagenes(ruta):
    img = cv2.imread(ruta) 
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    plt.figure(figsize=(18,5))

    plt.subplot(1,6,1)
    plt.imshow(img_rgb)
    plt.title("Color RGB")
    plt.axis("off")

    plt.subplot(1,6,2)
    plt.imshow(img_gray, cmap='gray')
    plt.title("Escala de Grises")
    plt.axis("off")

    plt.subplot(1,6,3)
    plt.imshow(img_rgb[:,:,0], cmap='gray')
    plt.title("Canal Rojo")
    plt.axis("off")

    plt.subplot(1,6,4)
    plt.imshow(img_rgb[:,:,1], cmap='gray')
    plt.title("Canal Verde")
    plt.axis("off")

    plt.subplot(1,6,5)
    plt.imshow(img_rgb[:,:,2], cmap='gray')
    plt.title("Canal Azul")
    plt.axis("off")

    plt.subplot(1,6,6)
    plt.imshow(img_gray, cmap='winter')
    plt.title("Color Falso")
    plt.axis("off")

    plt.tight_layout()
    plt.show()

    plt.figure(figsize=(12,7))

    plt.subplot(2,3,1)
    plt.imshow(img_rgb)
    plt.title("Color RGB")
    plt.axis("off")

    plt.subplot(2,3,2)
    plt.imshow(img_gray, cmap='gray')
    plt.title("Escala de Grises")
    plt.axis("off")

    plt.subplot(2,3,3)
    plt.imshow(img_rgb[:,:,0], cmap='gray')
    plt.title("Canal Rojo")
    plt.axis("off")

    plt.subplot(2,3,4)
    plt.imshow(img_rgb[:,:,1], cmap='gray')
    plt.title("Canal Verde")
    plt.axis("off")

    plt.subplot(2,3,5)
    plt.imshow(img_rgb[:,:,2], cmap='gray')
    plt.title("Canal Azul")
    plt.axis("off")

    plt.subplot(2,3,6)
    plt.imshow(img_gray, cmap='jet')
    plt.title("Falso Color")
    plt.axis("off")

    plt.tight_layout()
    plt.show()
```

Función con imagen 1

```python
mostrar_imagenes('foto-1.webp')
```

Función con imagen 2

```python
mostrar_imagenes('foto-2.webp')
```
