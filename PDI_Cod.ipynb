Ploteo por canales

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('colores.png', cv2.IMREAD_COLOR)

cmaps = ['gray','hot','cool','jet','hsv','spring','summer','autumn','winter','plasma']

plt.figure(figsize=(18, 6))

# Imagen original
plt.subplot(2, 6, 1)
plt.title('Original')
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.axis('off')

# 10 colores con loop
for i, cmap in enumerate(cmaps):
    plt.subplot(2, 6, i + 2)
    plt.title(cmap)
    plt.imshow(img[:,:,0], cmap=cmap)
    plt.axis('off')

plt.show()
```

Con colores relleno

```python
import cv2
import matplotlib.pyplot as plt  
import numpy as np

plt.rcParams['figure.figsize'] = [6, 4]

img = cv2.imread('colores.png', cv2.IMREAD_COLOR)
imgRGB = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

r = img[:,:,2]
g = img[:,:,1]
b = img[:,:,0]

# Crear imagen por canal 
r_img = np.zeros_like(imgRGB)
r_img[:,:,0] = r  

g_img = np.zeros_like(imgRGB)
g_img[:,:,1] = g 
b_img = np.zeros_like(imgRGB)
b_img[:,:,2] = b  

plt.figure(figsize=(6, 4))

plt.subplot(141)
plt.title('Original')
plt.imshow(imgRGB)
plt.axis('off')

plt.subplot(142)
plt.title('Rojo')
plt.imshow(r_img)
plt.axis('off')

plt.subplot(143)
plt.title('Verde')
plt.imshow(g_img)
plt.axis('off')

plt.subplot(144)         
plt.title('Azul')
plt.imshow(b_img)
plt.axis('off')

plt.tight_layout()
plt.show()
```

Imagenes con falso color

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('colores2.jpg', cv2.IMREAD_COLOR)
imgRGB = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
gray  = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# 10 colormaps
cmaps = ['gray', 'hot', 'cool', 'jet', 'hsv',
         'spring', 'summer', 'autumn', 'winter', 'plasma']

plt.figure(figsize=(18, 6))

# Imagen original 
plt.subplot(2, 6, 1)
plt.title('Original')
plt.imshow(imgRGB)
plt.axis('off')

# 10 colores 
for i, cmap in enumerate(cmaps):
    plt.subplot(2, 6, i + 2)
    plt.title(cmap)
    plt.imshow(gray, cmap=cmap)
    plt.axis('off')

plt.tight_layout()
plt.show()
```

# Ejercicios de Lógica

```python
# 1. Factorial
num = int(input("Ingrese un numero entero: "))
factorial = 1
if num < 0:
    print("No existe factorial de negativos")
elif num == 0:
    print("El factorial es 1")
else:
    for i in range(1, num + 1):
        factorial = factorial * i
    print("El factorial de", num, "es", factorial)
```

```python
# 2. Numero Primo
num = int(input("Ingrese un numero: "))
if num > 1:
    es_primo = True
    for i in range(2, num):
        if (num % i) == 0:
            es_primo = False
            break
    if es_primo:
        print(num, "es un numero primo")
    else:
        print(num, "no es un numero primo")
else:
    print(num, "no es un numero primo")
```

```python
# 3. Mayor y Menor
limite = int(input("Ingrese la cantidad de numeros (minimo 3): "))
if limite < 3:
    print("El limite debe ser mayor o igual a 3")
else:
    numeros = []
    for i in range(limite):
        n = float(input(f"Ingrese numero {i+1}: "))
        numeros.append(n)
        
    mayor = numeros[0]
    menor = numeros[0]
    
    for x in numeros:
        if x > mayor:
            mayor = x
        if x < menor:
            menor = x
            
    print("El mayor es:", mayor)
    print("El menor es:", menor)
```

```python
# 4. Palindromo
palabra = input("Ingrese una palabra: ")
palabra = palabra.lower().replace(" ", "")
invertida = palabra[::-1]
if palabra == invertida:
    print("Es un palindromo")
else:
    print("No es un palindromo")
```

```python
# 5. Numero mas repetido
numeros = []
print("Ingrese 7 numeros:")
for i in range(7):
    n = int(input(f"Numero {i+1}: "))
    numeros.append(n)

mas_repetido = numeros[0]
max_veces = 0

for i in numeros:
    veces = numeros.count(i)
    if veces > max_veces:
        max_veces = veces
        mas_repetido = i

print(f"El numero {mas_repetido} se repite {max_veces} veces")
```

```python
# 6. Anagrama
palabra1 = input("Ingrese la primera palabra: ").lower()
palabra2 = input("Ingrese la segunda palabra: ").lower()

lista1 = list(palabra1)
lista2 = list(palabra2)

lista1.sort()
lista2.sort()

if lista1 == lista2:
    print("Es un anagrama")
else:
    print("No es un anagrama")
```

```python
# 7. Metodo Burbuja
cantidad = int(input("Ingrese la cantidad de numeros a ordenar: "))
lista = []

for i in range(cantidad):
    num = float(input(f"Ingrese numero {i+1}: "))
    lista.append(num)

print("Lista original:", lista)

n = len(lista)
for i in range(n):
    for j in range(0, n-i-1):
        if lista[j] > lista[j+1]:
            lista[j], lista[j+1] = lista[j+1], lista[j]

print("Lista ordenada:", lista)
```
