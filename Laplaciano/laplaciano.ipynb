```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Cargar imagen
img = cv2.imread('suner.jpeg')
img_gris = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Aplicar Laplaciano
laplaciano = cv2.Laplacian(img_gris, cv2.CV_64F)
laplaciano_abs = np.uint8(np.absolute(laplaciano))

# Mostrar resultado
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.imshow(img_gris, cmap='gray')
plt.title('Imagen Original')
plt.axis('off')
plt.subplot(1, 2, 2)
plt.imshow(laplaciano_abs, cmap='gray')
plt.title('Laplaciano')
plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Cargar imagen
img = cv2.imread('suner.jpeg')
img_gris = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Aplicar filtro Gaussiano
gaussiano = cv2.GaussianBlur(img_gris, (5, 5), 1.4)

# Aplicar Laplaciano sobre la imagen suavizada
log = cv2.Laplacian(gaussiano, cv2.CV_64F)
log_abs = np.uint8(np.absolute(log))

# Mostrar resultado
plt.figure(figsize=(18, 5))
plt.subplot(1, 3, 1)
plt.imshow(img_gris, cmap='gray')
plt.title('Imagen Original')
plt.axis('off')
plt.subplot(1, 3, 2)
plt.imshow(gaussiano, cmap='gray')
plt.title('Gaussiano aplicado')
plt.axis('off')
plt.subplot(1, 3, 3)
plt.imshow(log_abs, cmap='gray')
plt.title('Laplaciano del Gaussiano (LoG)')
plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Cargar imagen
img = cv2.imread('suner.jpeg')
img_gris = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Laplaciano
laplaciano = cv2.Laplacian(img_gris, cv2.CV_64F)
laplaciano_abs = np.uint8(np.absolute(laplaciano))

# Laplaciano del Gaussiano
gaussiano = cv2.GaussianBlur(img_gris, (5, 5), 1.5)
log = cv2.Laplacian(gaussiano, cv2.CV_64F)
log_abs = np.uint8(np.absolute(log))

# Comparacion
plt.figure(figsize=(18, 5))
plt.subplot(1, 3, 1)
plt.imshow(img_gris, cmap='gray')
plt.title('Original')
plt.axis('off')
plt.subplot(1, 3, 2)
plt.imshow(laplaciano_abs, cmap='gray')
plt.title('Laplaciano')
plt.axis('off')
plt.subplot(1, 3, 3)
plt.imshow(log_abs, cmap='gray')
plt.title('Laplaciano del Gaussiano (LoG)')
plt.axis('off')
plt.tight_layout()
plt.show()
```
