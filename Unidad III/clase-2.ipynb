```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('colores2.jpg')

# Matriz Enfoque
kernel_enfoque = np.array([[0, -1, 0],
                           [-1, 5, -1],
                           [0, -1, 0]], np.float32)

img_enfoque = cv2.filter2D(img, -1, kernel_enfoque)


# Matriz Desenfoque
kernel_desenfoque = np.ones((3,3), np.float32) / 9

img_desenfoque = cv2.filter2D(img, -1, kernel_desenfoque)


# Realce de bordes
kernel_realce = np.array([[0, 0, 0],
                          [-1, 1, 0],
                          [0, 0, 0]], np.float32)

img_realce = cv2.filter2D(img, -1, kernel_realce)


# Matriz Repujado
kernel_repujado = np.array([[-2, -1, 0],
                            [-1, 1, 1],
                            [0, 1, 2]], np.float32)

img_repujado = cv2.filter2D(img, -1, kernel_repujado)


# Detección de bordes
kernel_bordes = np.array([[0, 1, 0],
                          [1, -4, 1],
                          [0, 1, 0]], np.float32)

img_bordes = cv2.filter2D(img, -1, kernel_bordes)


# Filtro tipo Sobel
kernel_sobel = np.array([[-1, 0, 1],
                         [-2, 0, 2],
                         [-1, 0, 1]], np.float32)

img_sobel = cv2.filter2D(img, -1, kernel_sobel)



# Filtro tipo Sharpen
kernel_sharpen = np.array([[1, -2, 1],
                           [-2, 5, -2],
                           [1, -2, 1]], np.float32)

img_sharpen = cv2.filter2D(img, -1, kernel_sharpen)


# Filtro Norte
kernel_norte = np.array([[1, 1, 1],
                         [1, -2, 1],
                         [-1, -1, -1]], np.float32)

img_norte = cv2.filter2D(img, -1, kernel_norte)


# Filtro Este
kernel_este = np.array([[-1, 1, 1],
                        [-1, -2, 1],
                        [-1, 1, 1]], np.float32)

img_este = cv2.filter2D(img, -1, kernel_este)


# Filtro Gauss
kernel_gauss = np.array([[1, 2, 3, 2, 1],
                         [2, 7, 11, 7, 2],
                         [3, 11, 17, 11, 3],
                         [2, 7, 11, 7, 2],
                         [1, 2, 3, 2, 1]], np.float32) / 159

img_gauss = cv2.filter2D(img, -1, kernel_gauss)


plt.figure(figsize=(15,10))

plt.subplot(3,4,1)
plt.title("Original")
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))

plt.subplot(3,4,2)
plt.title("Enfoque")
plt.imshow(cv2.cvtColor(img_enfoque, cv2.COLOR_BGR2RGB))

plt.subplot(3,4,3)
plt.title("Desenfoque")
plt.imshow(cv2.cvtColor(img_desenfoque, cv2.COLOR_BGR2RGB))

plt.subplot(3,4,4)
plt.title("Realce")
plt.imshow(cv2.cvtColor(img_realce, cv2.COLOR_BGR2RGB))

plt.subplot(3,4,5)
plt.title("Repujado")
plt.imshow(cv2.cvtColor(img_repujado, cv2.COLOR_BGR2RGB))

plt.subplot(3,4,6)
plt.title("Bordes")
plt.imshow(cv2.cvtColor(img_bordes, cv2.COLOR_BGR2RGB))

plt.subplot(3,4,7)
plt.title("Sobel")
plt.imshow(cv2.cvtColor(img_sobel, cv2.COLOR_BGR2RGB))

plt.subplot(3,4,8)
plt.title("Sharpen")
plt.imshow(cv2.cvtColor(img_sharpen, cv2.COLOR_BGR2RGB))

plt.subplot(3,4,9)
plt.title("Norte")
plt.imshow(cv2.cvtColor(img_norte, cv2.COLOR_BGR2RGB))

plt.subplot(3,4,10)
plt.title("Este")
plt.imshow(cv2.cvtColor(img_este, cv2.COLOR_BGR2RGB))

plt.subplot(3,4,11)
plt.title("Gauss")
plt.imshow(cv2.cvtColor(img_gauss, cv2.COLOR_BGR2RGB))

plt.show()
```
