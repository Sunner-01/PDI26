```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
def contraste(img,factor=1.0):
    img=img.astype(np.float32)
    media=np.mean(img)
    img_contraste=media+(img-media)*factor
    img_contraste=np.clip(img_contraste,0,255).astype(np.uint8)
    return img_contraste
img=cv2.imread('pego.png',0)
contrast=7
img_fin=contraste(img,contrast)
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(img,cmap='gray')
plt.title('original')
plt.subplot(1,2,2)
plt.imshow(img_fin,cmap='gray')
plt.title('contraste')
plt.show();
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
def contraste(img,factor=1.0):
    img=img.astype(np.float32)
    for i in range(3):
        media=np.mean(img[:,:,i])
        img[:,:,i]=media+(img[:,:,i]-media)*factor
        img[:,:,i]=np.clip(img[:,:,i],0,255)
    return img.astype(np.uint8)
img=cv2.imread('pego.png')
contrast=9
img_fin=contraste(img,contrast)

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
plt.title('original')
plt.subplot(1,2,2)
plt.imshow(cv2.cvtColor(img_fin,cv2.COLOR_BGR2RGB))
plt.title('contraste')
plt.show();
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
def brillo(img,brillo=0):
    img_brillo=img.astype(np.float32)+brillo
    img_brillo=np.clip(img_brillo,0,255).astype(np.uint8)
    return img_brillo
img=cv2.imread('pego.png',0)
bril=-70
img_fin=brillo(img,bril)
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(img,cmap='gray')
plt.subplot(1,2,2)
plt.imshow(img_fin,cmap='gray')
plt.title('brillo')
plt.show();
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
def brillo_color(img, brillo=0):
    img_brillo = img.astype(np.float32) + brillo
    img_brillo = np.clip(img_brillo, 0, 255)
    return img_brillo.astype(np.uint8)
img = cv2.imread('pego.png')
bril = -70
img_fin = brillo_color(img, bril)
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('original')
plt.subplot(1,2,2)
plt.imshow(cv2.cvtColor(img_fin, cv2.COLOR_BGR2RGB))
plt.title('brillo color')

plt.show()
```

MASCARAS

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
img=cv2.imread('pego.png',0)
alto,ancho=img.shape[:2]
mascara=np.zeros_like(img)
centro=(ancho//2,alto//2)
radio=70
cv2.circle(mascara,centro,radio,(255),-1)
img_mascara=cv2.bitwise_and(img,img,mask=mascara)
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(img,cmap='gray')
plt.title('original')
plt.subplot(1,2,2)
plt.imshow(img_mascara,cmap='gray')
plt.title('mascara')
plt.show();
```
